```python
# Implementasi Hash Table Sederhana

class SimpleHashTable:
  def __init__(self,size):
    self.size = size
    self.table =[None] * size

  def hash_function(self,key):
    return hash(key) % self.size

  def insert(self,key,value):
    index = self.hash_function(key)
    self.table[index] = value

  def search(self,key):
    index = self.hash_function(key)
    return self.table[index]

# Contoh Penggunaan
ht = SimpleHashTable(10)
ht.insert("apel", 100)
ht.insert("jeruk", 259)

print("nilai untuk 'apel':", ht.search("apel"))
print("nilai untuk 'jeruk':", ht.search("jeruk"))
```

```python
# Penanganan Collison dengan Linear Probing

class LinearProbingHashTable:
  def __init__(self,size):
    self.size = size
    self.table =[None] * size

  def hash_function(self,key):
    return hash(key) % self.size
  def insert(self,key,value):
    index =self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index][0] == key:
        self.table[index] = (key,value)
        break
      index = (index + 1) % self.size
      if index == original_index:
        raise Exception("Hash table penuh")
    self.table[index] = (key,value)

  def search(self,key):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table [index][0] == key:
        return self.table[index][1]
      index = (index + 1) % self.size
      if index == original_index:
        return None
    return None

#Contoh Penggunaan
ht = LinearProbingHashTable(10)
ht.insert("apel", 100)
ht.insert("jeruk", 259)
ht.insert("mangga", 300)
print("nilai untuk 'apel':", ht.search("apel"))
print("nilai untuk 'mangga':", ht.search("mangga"))
```

```python
# Penanganan Collison dengan Chaining

class ChainingHashTable:
  def __init__(self,size):
    self.size = size
    self.table = [[] for _ in range(size)]
  def hash_function(self,key):
    return hash(key) % self.size
  def insert (self,key,value):
    index = self.hash_function(key)
    for i, (k, v) in enumerate(self.table[index]):
      if k == key:
        self.table[index][i] = (key,value)
        return
    self.table[index].append((key,value))
  def search(self,key):
    index = self.hash_function(key)
    for k, v in self.table[index]:
      if k == key:
        return v
    return None

#Contoh Penggunaan
ht = ChainingHashTable(10)
ht.insert("apel", 100)
ht.insert("jeruk", 259)
ht.insert("mangga", 300)
print("nilai untuk 'apel':", ht.search("apel"))
print("nilai untuk 'mangga':", ht.search("mangga"))
```

```python
# Tugas 1
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return ord(key[0]) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                break
            index = (index + 1) % self.size
            if index == original_index:
                raise Exception("Hash table penuh")

        self.table[index] = (key, value)

    def print_table(self):
        print("Isi Hash Table:")
        for i, slot in enumerate(self.table):
            print(f"Index {i}: {slot}")

# a. Buat objek dengan ukuran tabel 5
ht_linear = LinearProbingHashTable(5)

# b. Masukkan pasangan key-value sesuai soal
ht_linear.insert("A", 10)
ht_linear.insert("B", 20)
ht_linear.insert("C", 30)
ht_linear.insert("D", 40)
ht_linear.insert("E", 50)

# c. Cetak isi tabel
ht_linear.print_table()
```

```python
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = []

        for _ in range(size):
            self.table.append([])

    # Fungsi hash sederhana
    def hash_function(self, key):
        return len(key) % self.size

    # Menambahkan data ke hash table
    def insert(self, key, value):
        posisi = self.hash_function(key)
        self.table[posisi].append((key, value))

    # Menampilkan isi hash table
    def display(self):
        print("Isi Hash Table:")
        for nomor, bucket in enumerate(self.table):
            print(f"Index {nomor}: {bucket}")


# a. Gunakan ukuran tabel 3
hash_table = ChainingHashTable(3)

# b. Masukkan key-value sesuai instruksi soal
hash_table.insert("apel", 100)
hash_table.insert("melon", 200)
hash_table.insert("lemon", 300)
hash_table.insert("pisang", 400)

# c. Cetak isi tabel
hash_table.display()
```
