```python
# MEMBUAT GRAPH TAK BERARAH

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B', 'G'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
    'G': ['D']
}

print("=== GRAPH TAK BERARAH ===")

for node in graph:
    print(node, "->", graph[node])
```

```python
# implementasi BFS dan DFS

from collections import deque

#BFS
def bfs(graph, start):

    visited = set()
    queue = deque([start])

    while queue:

        vertex = queue.popleft()

        if vertex not in visited:

            print(vertex, end=" ")
            visited.add(vertex)

            queue.extend([
                neighbor for neighbor in graph[vertex]
                if neighbor not in visited
            ])

print("=== HASIL BFS ===")
bfs(graph, 'A')

#DFS
def dfs(graph, start, visited=None):

    if visited is None:
        visited = set()

    print(start, end=" ")
    visited.add(start)

    for neighbor in graph[start]:

        if neighbor not in visited:
            dfs(graph, neighbor, visited)

print("\n\n=== HASIL DFS ===")
dfs(graph, 'A')
```

```python
# Modifikasi BFS Agar Mengembalikan List Urutan Kunjungan
from collections import deque

def bfs(graph, start):

    visited = set()
    queue = deque([start])
    hasil = []

    while queue:

        vertex = queue.popleft()

        if vertex not in visited:

            visited.add(vertex)
            hasil.append(vertex)

            queue.extend([
                neighbor for neighbor in graph[vertex]
                if neighbor not in visited
            ])

    return hasil

print("=== BFS MODIFIKASI ===")
print(bfs(graph, 'A'))
```

```python
# Fungsi path
def path(graph, start, end, visited=None):

    if visited is None:
        visited = set()

    visited.add(start)

    if start == end:
        return [start]

    for neighbor in graph[start]:

        if neighbor not in visited:

            jalur = path(graph, neighbor, end, visited)

            if jalur:
                return [start] + jalur

    return None

print("=== PATH A KE G ===")
print(path(graph, 'A', 'G'))
```

```python
# Fungsi is_connected(graph)
def is_connected(graph):

    visited = set()

    start = next(iter(graph))

    def dfs_connected(vertex):

        visited.add(vertex)

        for neighbor in graph[vertex]:

            if neighbor not in visited:
                dfs_connected(neighbor)

    dfs_connected(start)

    return len(visited) == len(graph)

print("=== CEK CONNECTED GRAPH ===")
print(is_connected(graph))
```
